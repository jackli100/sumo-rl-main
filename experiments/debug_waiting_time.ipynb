```python
import pandas as pd
import matplotlib.pyplot as plt

# 读取CSV文件
csv_file_path = r'D:\trg1vr\sumo-rl-main\sumo-rl-main\dqn_conn0_ep1 (3).csv'  # 替换为你本地文件的实际路径
data = pd.read_csv(csv_file_path)

# 显示前几行数据以了解数据结构
print(data.head())

# 绘制系统总等待时间和代理总等待时间的折线图
plt.figure(figsize=(12, 6))

# 假设CSV文件中包含'system_total_waiting_time'和'agents_total_accumulated_waiting_time'列
plt.plot(data['system_total_waiting_time'], label='System Total Waiting Time')
plt.plot(data['agents_total_accumulated_waiting_time'], label='Agent Total Accumulated Waiting Time')

plt.xlabel('Steps')
plt.ylabel('Waiting Time')
plt.title('System vs. Agent Total Waiting Time')
plt.legend()
plt.grid(True)
plt.show()
```
